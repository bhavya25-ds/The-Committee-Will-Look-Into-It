```python
import pandas as pd

df = pd.read_csv('paper_leaks.csv')

# basic shape
print(df.shape)
print("\n")

# column types
print(df.dtypes)
print("\n")

# missing values
print(df.isnull().sum())
print("\n")

# unique values in categorical columns
for col in ['era', 'body_type', 'area', 'leak_status', 'action_taken', 'confidence']:
    print(f"{col}: {df[col].unique()}")
    print()

# date range
print(df['date'].min(), "to", df['date'].max())

# aspirants affected — is it numeric or string?
print("\naspirants_affected dtype:", df['aspirants_affected'].dtype)
print(df['aspirants_affected'].head(10))
```
